Titanic Dataset: Data Exploration and Cleaning with Pandas

1. Import the required library

```python
import pandas as pd
```

2. Load the dataset

```python
file_path = "Titanic-3CS03.csv"
df = pd.read_csv(file_path)
print("Dataset loaded successfully.")
print("Number of rows: {df.shape[0]}")
print("Number of columns: {df.shape[1]}")
```

3. Inspect the data

3.1 Display the full dataset

```python
display(df)
```

3.2 View the first and last rows

```python
display(df.head())
display(df.tail())
```

3.3 View column names

```python
print("Column names:")
print(df.columns.tolist())
```

3.4 Check dataset dimensions

```python
rows, columns = df.shape
print(f"Rows: {rows}")
print(f"Columns: {columns}")
```

3.5 Check data types

```python
print(df.dtypes)
```

3.6 Get a compact structural summary

```python
df.info()
```

3.7 Summary statistics

```python
display(df.describe())
```

4. Identify missing values

4.1 Display missing-value locations

```python
display(df.isna())
```

4.2 Count missing values in each column

```python
missing_before = df.isna().sum().sort_values(ascending=False)
print("Missing values in each column:")
display(missing_before.to_frame(name="Missing count"))
```

5. Handle missing values

```python
# Fill missing Age values using the median age.
if "Age" in df.columns:
  age_median = df["Age"].median()
  df["Age"] = df["Age"].fillna(age_median)
  print(f"Missing Age values filled with median: {age_median}")
# Fill missing Fare values using the mean fare.
if "Fare" in df.columns:
  fare_mean = df["Fare"].mean()
  df["Fare"] =df["Fare"].fillna(fare_mean)
  print(f"Missing Fare values filled with mean: {fare_mean}")
# Fill missing Embarked values using the most frequent category.
if "Embarked" in df.columns:
  embarked_mode = df["Embarked"].mode(dropna=True)
  if not embarked_mode.empty:
    df["Embarked"] = df["Embarked"].fillna(embarked_mode.iloc[0])
    print(f"Missing Embarked values filled with mode: {embarked_mode.iloc[0]}")
    display(df.head())
```

5.1 Check missing values again

```python
missing_after_imputation = df.isna().sum().sort_values(ascending=False)
display(missing_after_imputation.to_frame(name="Missing count"))
```

6. Find and remove duplicate rows

```python
duplicate_count = df.duplicated().sum()
print(f"Exact duplicate rows found: {duplicate_count}")
df = df.drop_duplicates().copy()
print(f"Rows remaining after duplicate removal: {len(df)}")
```

7. Decide how to handle any remaining missing values

```python
print("Rows before optional complete-case filtering:", len(df))
df_complete = df.dropna().copy()
print("Rows if all rows with any missing value were removed:", len(df_complete))
print("Rows that would be removed:", len(df) - len(df_complete))
# We continue with df (not df_complete) so that we do not silently discard rows.
```

8. Convert categorical columns into numeric columns

```python
categorical_columns = [col for col in ["Sex", "Embarked"] if col in df.columns]
if categorical_columns:
  df = pd.get_dummies(
      df,
      columns=categorical_columns,
      dtype=int
  )
  print("Encoded categorical columns:", categorical_columns)
else:
  print("No requested categorical columns were found; nothing to encode.")
  display(df.head())
```

9. Validate the cleaned dataset

```python
print("cleaned dataset preview:")
display(df.head())
print("\nFinal shape (rows, columns):", df.shape)
print("\nMissing values by column:")
display(df.isna().sum().to_frame(name="Missing count"))
print("\nExact duplicate rows remaining:", df.duplicated().sum())
```

10. Save the cleaned dataset

```python
output_path = "titanic_3CS03-Cleaned.csv"
df.to_csv(output_path, index=False)
print(f"Data cleaning completed. Cleaned file saved as: {output_path}")
```

Program 10: First Data Science Program Using Pandas

```python
# Step 1: Import pandas
import pandas as pd
# Step 2: Create data
data = {
    "Name": ["Amit", "Ravi", "Priya", "Rahul",],
    "Marks": [80, 75, 90, 85]
}
# Step 3: Create DataFrame
df = pd.DataFrame(data)
# Step 4: Display the data
print("Student Data:")
print(df)
# Step 5: Calculate average
print("\nAverage Marks =", df["Marks"].mean())
```

Program 18: Count Even and Odd Numbers in a List

```python
numbers = [10, 15, 20, 25, 30, 35]
even_count = 0
odd_count = 0
for num in numbers:
  if num % 2 == 0:
    even_count += 1
  else:
    odd_count += 1

print("Even numbers =", even_count)
print("Odd numbers =", odd_count)
```

Program 19: Create a Simple Bar Chart

```python
import matplotlib.pyplot as plt
subjects = ["Python", "DBMS", "OS", "CN"]
marks = [85, 78, 90, 82]
plt.bar(subjects, marks)
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

Experiment 7: Hypothesis Testing using a Two-Sample t-Test.

Dataset: California Housing Dataset.

Objective: Compare average housing values between two
household income categories.

# Step 1: Import the required libraries.

```python
import pandas as pd
# Pandas is used to organise and manipulate dataset values.

from scipy.stats import ttest_ind
# ttest_ind performs an independent two-sample t-test.

from sklearn.datasets import fetch_california_housing
# This function loads the California Housing dataset.
```

# Step 2: Load the California Housing dataset.

```python
housing = fetch_california_housing(as_frame=True)
# Loads the dataset as a Pandas DataFrame.
# The first call may download the dataset if it is not cached.

df = housing.frame
# Stores the dataset, including its features and target,
# in a DataFrame named df.
```

# Step 3: Display basic information about the dataset.

```python
print("Dataset Shape:", df.shape)
# Displays the number of rows and columns in the dataset.

print("\nFirst five rows of the dataset:")
# Prints a heading for the first five observations.

print(df.head())
# Displays the first five rows to understand the data.
```

# Step 4: Identify the two income categories.

```python
median_income = df["MedInc"].median()
# Calculates the median of MedInc (median household income).
# MedInc is measured in units of $10,000.

print("\nMedian Income Threshold:", median_income)
# Displays the threshold used to divide the income groups.

low_income = df[df["MedInc"] < median_income]["MedHouseVal"]
# Selects housing values for observations below the median income.
# MedHouseVal represents median house value in units of $100,000.

high_income = df[df["MedInc"] >= median_income]["MedHouseVal"]
# Selects housing values for observations at or above
# the median household income.
```

# Step 5: Calculate descriptive statistics

```python
print("\n--- Descriptive Statistics ---")
# Displays a heading for the summary statistics.

print("Low-income group size:", len(low_income))
# Displays the number of observations in the low-income group.

print("High-income group size:", len(high_income))
# Displays the number of observations in the high-income group.

print("Mean housing value (low income):", low_income.mean())
# Calculates the average housing value in the low-income group.

print("Mean housing value (high income):", high_income.mean())
# Calculates the average housing value in the high-income group.

print("Difference in means:", high_income.mean() - low_income.mean())
# Calculates the difference between the two sample means.
```

# Step 6: Formulate the hypotheses

```python
# Null hypothesis (H0):
# The population mean housing values are equal.
# H0: mu_low = mu_high

# Alternative hypothesis (H1):
# The population mean housing values are different.
# H1: mu_low != mu_high

print("\n--- Hypothesis Testing ---")
# Displays the hypothesis-testing section.

print("H0: Average housing values are equal.")
# States the null hypothesis.

print("H1: Average housing values are different.")
# States the alternative hypothesis.
```

# Step 7: Set the significance level

```python
alpha = 0.05
# Sets the significance level to 5%.
# This is the threshold for deciding statistical significance.

print("\nSignificance Level:", alpha)
# Displays the selected significance level.
```

# Step 8: Perform the independent two-sample t-test

```python
t_statistic, p_value = ttest_ind(
    low_income,
    high_income,
    equal_var=False,
    alternative="two-sided"
)
# Compares the two independent groups.
# equal_var=False selects Welch's t-test, which does not
# assume that the population variances are equal.
# alternative="two-sided" tests for a difference in either
# direction (higher or lower housing values).
```

# Step 9: Display the test results

```python
print("\n--- Test Results ---")
# Displays a heading for the results.

print("T-statistic:", t_statistic)
# Displays the calculated t-statistic.
# Its sign indicates the direction of the mean difference
# based on the order of the groups supplied to the test.

print("P-value:", p_value)
# Displays the p-value used to assess the evidence
# against the null hypothesis.
```

# Step 10: Interpret the p-value and make a decision

```python
if p_value < alpha:
  # Executes this block when the p-value is below 0.05.

  print("\nDecision: Reject the null hypothesis.")
  # Rejects H0 because the result is statistically significant.

  print("Conclusion: Average housing values differ")
  print("significantly between the two income categories.")
  # Reports evidence of a difference in population means.

else:
  # Executes this block when the p-value is 0.05 or greater.

  print("\nDecision: Fail to reject the null hypothesis.")
  # The evidence is insufficient to reject H0.

  print("Conclusion: There is insufficient evidence")
  print("of a difference in average housing values.")
  # Does not claim that the two population means are equal.
```

# Step 11: Display the final interpretation

```python
print("\n--- Final Interpretation ---")
# Prints the final summary heading.

if p_value < alpha:
  print("Since p-value < 0.05, the result is statistically significant.")
  # Explains why the null hypothesis is rejected.

else:
  print("Since p-value >= 0.05, the result is not statistically significant.")
  # Explains why the null hypothesis is not rejected.

  # END OF PROGRAM
  # ----------------------------------------------------------
```
